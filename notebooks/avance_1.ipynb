```python
from pathlib import Path
from urllib.request import urlopen

import pandas as pd

# Versión fija de los datos oficiales del curso.
# Esta versión coincide con los 10 archivos que ya están en Equipo4/datos/crudos.
ORIGEN = (
    "https://raw.githubusercontent.com/LuisRojo-Gonzalez/"
    "Operations-Management/4f2718dc185dd3a5e94243424e386b558a460178/"
    "CaseStudy/Datos"
)

DATO_DIR = Path('datos/crudos')
DATO_DIR.mkdir(parents=True, exist_ok=True)

ARCHIVOS = [
    'calidad_atributos.csv',
    'calidad_llenado.csv',
    'capacidad_horizonte.csv',
    'compromisos_semana_focal.csv',
    'demanda_semanal.csv',
    'especificaciones_calidad.csv',
    'inventario_inicial.csv',
    'maestro_sku.csv',
    'metadatos.json',
    'setups_cip.csv'
]

# Si los datos ya están en la carpeta, se usan tal cual.
# Si se abre desde cero en Colab, se descargan automáticamente.
for nombre in ARCHIVOS:
    destino = DATO_DIR / nombre
    if not destino.exists():
        with urlopen(f'{ORIGEN}/{nombre}', timeout=30) as respuesta:
            destino.write_bytes(respuesta.read())
    assert destino.is_file() and destino.stat().st_size > 0, f'Falta un archivo: {nombre}'

print(f'Datos listos: {len(ARCHIVOS)} archivos originales.')
```

```python
import json

# DataFrames disponibles para el análisis; no se alteran los CSV originales.
tablas = {
    Path(nombre).stem: pd.read_csv(DATO_DIR / nombre)
    for nombre in ARCHIVOS if nombre.endswith('.csv')
}
metadatos = json.loads((DATO_DIR / 'metadatos.json').read_text(encoding='utf-8'))

resumen = pd.DataFrame([
    {'archivo': nombre, 'filas': len(df), 'columnas': len(df.columns)}
    for nombre, df in tablas.items()
])
display(resumen)

# Vista preliminar de demanda semanal:
display(tablas['demanda_semanal'].head())
```
